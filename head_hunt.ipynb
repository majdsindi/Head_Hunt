##4: Using operators  

Here, we will be moving on to basic operators on values and variables. You can perform mathematical operations using the symbols you are familiar with (+,-, *, /, //,%).  Since we have defined `my_integer` by running the cell above, it will exist in memory (until we reset the kernel) and so we can use it in the future.
In the cell below, we multiply `my_integer` by 12 and store the result in a new variable called `result` .

```python
print("My integer is equal to",my_integer)

result=my_integer*12
print("My integer is equal to",my_integer)
print("Result is equal to",result)
```

Note that this operation does not change the value of `my_integer`.  Operating on a variable does not change the value of that variable.  If we want to store the result of an operation, we have to assign it to a variable - here we have assigned it to the aptly named variable `result`.

If we wanted to update the value of `my_integer` with the result, we could do so using the following code:

```python
print("My integer is equal to",my_integer)
my_integer=my_integer*12
print("My integer is equal to",my_integer)
```

If you run this cell multiple times then you will see that it gives a different output each time.  Why is that?  It is using the value of `my_integer` which was stored in the previous run of the cell.

There is also a slightly shorter way of writing the code to carry out a mathematical operation on a variable and then update that variable.  For example, we can rewrite the previous code as follows:

```python
print("My integer is equal to",my_integer)
my_integer*=12 #putting the * before the = is a shorthand way of writing my_integer=my_integer*12
print("My integer is equal to",my_integer)
```

##6: Using conditional Statements

Often when we program, we want different things to happen depending on the situation.  For example, the following line of code prints the statement, "In which city is the University of Sussex based?", waits for the user to input an answer and then stores this string in the variable `response`

```python
response=input("In which city is the University of Sussex based? ")
```

What the system should do next depends on the value stored in `response`.  We can check whether it is equivalent to a certain value using the `==` (double equals) operator.  

```python
correct_answer="Brighton" #single equals (=) stores the value of Brighton in a variable called correct_answer
response == correct_answer #double equals (==) compares the two values and outputs True if they are equivalent and False otherwise. It does not update response
```

Note that we get the Boolean value `True` if the user's response is equivalent to the correct response (and `False` otherwise).  We can combine this with an `if-then` construction to tell the user that they got it right.

```python
if response == correct_answer:
  print("Well done.  This is the correct answer")
```

Note the syntax that is used here.  The `if` part of the construction starts with the keyword `if` which is followed by the condition which we are testing and completed with a `:`.  On the next line, we indent the `then` part.  This is what should happen if the condition being tested is found to be `True`.  This can be multiple lines - but they all need to be indented!  Run the following cell multiple times.  What gets printed when the user gets the right answer and what gets printed when they get the wrong answer?

```python
response=input("In which city is the University of Sussex based? ")
if response == correct_answer:
  print("Well done.")
  print("This is the correct answer")
print("Goodbye")
```

Maybe we want to give the user another message if they got it wrong.  We could do this with another `if-then` construct where we use the != (not equals) operator in the condition: `if response != "Brighton"`



```python
response=input("In which city is the University of Sussex based? ")
if response == correct_answer:
  print("Well done.")
  print("This is the correct answer")
if response !=correct_answer:
  print("That is the incorrect answer")
  print("The correct answer is", correct_answer)
print("Goodbye")
```

However, it is much more normal to do this with a `if-then-else`

```python
response=input("In which city is the University of Sussex based? ")
if response == correct_answer:
  print("Well done.")
  print("This is the correct answer")
else:
  print("That is the incorrect answer")
  print("The correct answer is", correct_answer)
print("Goodbye")
```

We can test more complex conditions using the logical connectives `and`, `or` and `not`.  We can also use other logical comparisons which make sense for the datatypes being compared.  For example, we can use `<` (less than) and `>` (greater than) for numeric datatypes.

In our example,  we might want to accept multiple correct answers using an `or` as below:

```python
response=input("In which city is the University of Sussex based? (not case sensitive)")
almost_correct="brighton"
if response == correct_answer or response == almost_correct:
  print("Well done.")
  print("This is the correct answer")
else:
  print("That is the incorrect answer")
  print("The correct answer",correct_answer)
print("Goodbye")
```

But maybe we want to give a different message in each of these cases.
We can carry out multiple condition checks (with different consequents) using an `elif` clause.  This is short for else if and extends the if-then-else construction as below:

```python
response=input("In which city is the University of Sussex based? ")
if response == correct_answer:
  print("Well done.")
  print("This is the correct answer")
elif response == almost_correct:
  print("Well done - you have the correct city")
  print("However, in English the first letter of names of places should be capitalised")
else:
  print("That is the incorrect answer")
  print("The correct answer is",correct_answer)
print("Goodbye")
```

##7: Getting started with loops

Imagine that we want to print out every integer from 1-100 together with its square.  Naively, we could do this by extending the code in the cell below.


```python
number = 1
print(number," squared is ",number*number)
number = 2
print(number," squared is ",number*number)
number = 3
print(number," squared is ",number*number)

#and so on.  I don't really feel like writing (or even copy-pasting) the next 194 lines of code to finish this off tho!
```

Alternatively, we can do write this much more concisely using a loop or an iteration.

Loops are a way of repeating the same action multiple times without writing a line of code for each repetition.  This is really useful when we want to do 100, 1000 or 1,000,000+ repetitions of something.

There are two main types of loops we will look at, __for__ loops and __while__ loops.

__for__ loops iterate over a collection of items, e.g., a sequence of numbers. It will always execute a set number of times.   We will return to for loops next time when we look at lists and other collections.

__while__ loops repeat until a certain condition is met.  In the cell below, we use a __while__ loop to complete the squared numbers.

```python
number = 1 #initialise the variable on which the loop condition is going to depend
while number < 101:  # this is the loop condition.  The code inside the loop will keep repeating for as long as the condition is true.  For the loop to end, number must be changed to a number which is equal to or greater than 101
  print(number,"squared is ",number*number) #the code to be run on each repetition must be indented
  number+=1 #we have to update the loop variable - this means that the next number will be squared on the next repetition and at some point, the loop condition will not be met any more
print("Finished") #this code is not indented and will only be run after the loop has completed
```

A while loop will run whilst a given condition remains true. This is often used when we don't know how many times a loop needs to run.  In the example below, the loop will continue running until the user types exit


```python
inp=""
number = 1
while inp!="exit":
  print(number," squared is ",number*number)
  number+=1
  inp=input("Type 'next' for next square number or 'exit' to exit ")
```

```python
# Initialize counters for questions, correct answers, and incorrect answers
questions_asked = 0
correct_answers = 0
incorrect_answers = 0

# Loop through the times tables from 2 to 12
for i in range(2, 13):
  for j in range(1, 13):
    # Ask the question and get user input
    answer = input(f"What is {i} x {j}? ")

    # Increment the questions asked counter
    questions_asked += 1

    # Use try-except to handle non-numeric input
    try:
      # Check if the answer is correct
      if int(answer) == i * j:
        print("Correct answer!")
        correct_answers += 1  # Increment correct answers counter
      else:
        print(f"Incorrect answer. The correct answer is {i * j}")
        incorrect_answers += 1

        # Check if the user has made 3 incorrect attempts
        if incorrect_answers == 3:
          print(f"\nOn this attempt at the times table challenge, you completed {questions_asked} questions.")
          print(f"You might want to practice your {i} times table.")
          exit()  # Stop the program

    except ValueError:
      print("Invalid input. Please enter a number.")
      incorrect_answers += 1

      # Check if the user has made 3 incorrect attempts
      if incorrect_answers == 3:
        print(f"\nOn this attempt at the times table challenge, you completed {questions_asked} questions.")
        if questions_asked < 12:
          print(f"You might want to practice your {i} times table.")
        exit()  # Stop the program

# Display the final results if the user didn't make 3 incorrect attempts
print("\nTimes Table Challenge Completed!")
print(f"Total questions asked: {questions_asked}")
print(f"Total correct answers: {correct_answers}")
```
